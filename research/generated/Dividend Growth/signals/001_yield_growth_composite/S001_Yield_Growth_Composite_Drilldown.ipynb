## 1 · Signal Construction — Raw Inputs for This Ticker

Recomputes yield and growth directly from the DB (same logic as `build_signals.py`) for `TICKER` only, step by step.

```python
import psycopg2

DB             = dict(host="localhost", dbname="sfera", user="postgres", password="lokomotiv")
YIELD_CAP      = 0.15
GROWTH_CLIP_LO = -1.0
GROWTH_CLIP_HI =  5.0

# ── Step A: raw dividend payments from DB ────────────────────────────────────
conn = psycopg2.connect(**DB)
divs_tk = pd.read_sql(
    """
    SELECT ex_date, dividend
    FROM eodhd.dividends
    WHERE exchange = 'LSE'
      AND ticker   = %(ticker)s
      AND ex_date IS NOT NULL
      AND dividend > 0
    ORDER BY ex_date
    """,
    conn,
    params={"ticker": TICKER},
    parse_dates=["ex_date"],
)
conn.close()

divs_tk["dividend"] *= 100   # GBP → GBX
divs_tk["ex_date"]  = pd.to_datetime(divs_tk["ex_date"]).dt.normalize()
divs_tk = divs_tk.set_index("ex_date")["dividend"]

gaps = pd.Series(divs_tk.index).diff().dt.days.dropna()
median_gap = gaps.median()
typical_freq = "quarterly" if median_gap < 100 else "semi-annual" if median_gap < 200 else "annual"

# ── Step B: rolling TTM variants (window slider) ──────────────────────────────
price_idx_naive = t_price.index.normalize()
cal_start = price_idx_naive.min() - pd.Timedelta(days=600)
all_cal   = pd.date_range(start=cal_start, end=price_idx_naive.max(), freq="D")
div_daily = divs_tk.reindex(all_cal, fill_value=0.0).fillna(0.0)

WINDOWS = [252, 260, 280, 300, 365, 400, 450, 500, 550]
ttm_variants: dict[str, pd.Series] = {}
for w in WINDOWS:
    rolled = div_daily.rolling(w).sum()
    on_td  = rolled.reindex(price_idx_naive).copy()
    on_td.index = t_price.index
    ttm_variants[f"{w}d"] = on_td

# 12-month calendar variant
cal_ttm = pd.Series(np.nan, index=price_idx_naive, dtype=float)
for i, dt in enumerate(price_idx_naive):
    cutoff = dt - pd.DateOffset(months=12)
    cal_ttm.iloc[i] = divs_tk[(divs_tk.index > cutoff) & (divs_tk.index <= dt)].sum()
cal_ttm.index = t_price.index
ttm_variants["12-mo calendar"] = cal_ttm

# Last-pay ffill base
pay_dates_in_range = divs_tk.index[divs_tk.index >= price_idx_naive.min() - pd.Timedelta(days=400)]
ttm_at_pay = pd.Series(dtype=float)
for ex_date in pay_dates_in_range:
    cutoff = ex_date - pd.DateOffset(months=12)
    ttm_at_pay[ex_date] = divs_tk[(divs_tk.index > cutoff) & (divs_tk.index <= ex_date)].sum()
ttm_ffill_naive = ttm_at_pay.reindex(price_idx_naive, method="ffill")
ttm_ffill = ttm_ffill_naive.copy()
ttm_ffill.index = t_price.index
ttm_variants["last-pay ffill"] = ttm_ffill

# ── Step C: EWMA smoothing variants (base = last-pay ffill) ──────────────────
ewma_variants: dict[str, pd.Series] = {"no smoothing": ttm_ffill}
for span in [10, 20, 30, 40, 60, 90, 120]:
    ewma_variants[f"EWMA span={span}d"] = ttm_ffill.ewm(span=span, adjust=False).mean()

# Canonical 365d series used downstream
ttm_td      = ttm_variants["365d"]
ttm_prev_td = div_daily.rolling(365).sum().shift(365).reindex(price_idx_naive).copy()
ttm_prev_td.index = t_price.index
is_payer    = ttm_td > 0.001

# ── Charts ────────────────────────────────────────────────────────────────────
div_in_range = divs_tk[divs_tk.index >= price_idx_naive.min()]
raw_df  = pd.DataFrame({"time": div_in_range.index, "value": div_in_range.values})
ttm_df  = pd.DataFrame({"time": ttm_td.index,       "value": ttm_td.values})
ffil_df = pd.DataFrame({"time": ttm_ffill.index,    "value": ttm_ffill.values})

raw_chart = Chart(height=110).histogram(raw_df, name="Dividend payment (GBX)", color="#90caf9")

window_chart = (
    Chart(height=150)
    .area(ttm_df, name="TTM (365d)", color="#ffa726")
    .smoothing_control(variants=ttm_variants, label="Window", color="#ffa726", variants_init="365d")
)

smooth_chart = (
    Chart(height=150)
    .area(ffil_df, name="last-pay ffill", color="#ab47bc")
    .smoothing_control(variants=ewma_variants, label="Smoothing", color="#ab47bc", variants_init="EWMA span=90d")
)

Dashboard(
    panes=[raw_chart, window_chart, smooth_chart],
    titles=[
        f"{TICKER} — Raw Payments (GBX)  ·  {len(divs_tk)} payments  ·  ~{typical_freq} ({median_gap:.0f}d gap)",
        f"{TICKER} — Step B: TTM rolling window (slider)",
        f"{TICKER} — Step C: EWMA-smoothed TTM on last-pay ffill (slider, default 90d)",
    ],
    theme="dark",
)
```

```python
# -- Step D: Trailing yield % = two fixed lines, no slider --
EWMA_SPAN = 90

ttm_smooth    = ewma_variants[f'EWMA span={EWMA_SPAN}d']
is_pay_smooth = ttm_smooth > 0.001

yield_smooth  = (ttm_smooth / t_price).where(is_pay_smooth).clip(upper=YIELD_CAP) * 100
yield_raw_365 = (ttm_variants['365d'] / t_price).where(is_payer).clip(upper=YIELD_CAP) * 100

yld_raw_df = pd.DataFrame({'time': yield_raw_365.index, 'value': yield_raw_365.values})
yld_smo_df = pd.DataFrame({'time': yield_smooth.index,  'value': yield_smooth.values})

Dashboard(
    panes=[
        Chart(height=200)
        .line(yld_raw_df, name='Yield % 365d raw',            color='#ffa726')
        .line(yld_smo_df, name=f'Yield % EWMA-{EWMA_SPAN}d', color='#ab47bc')
    ],
    titles=[f'{TICKER} - Step D: Yield  orange=365d raw  purple=EWMA-{EWMA_SPAN}d smoothed'],
    theme='dark',
)
```

### 1b · Signal Components — Trailing Yield % and YoY Dividend Growth %

> **S001 has two components only:**
> - **40% — Trailing yield rank** = TTM dividend GBX / price GBX, ranked cross-sectionally
> - **60% — YoY dividend growth rank** = (TTM_now / TTM_prev_year) − 1, ranked cross-sectionally
>
> There is **no EPS component** in S001. EPS growth is used in S003.

```python

# -- Component 2: YoY dividend growth % --
all_cal_c   = pd.date_range(start=price_idx_naive.min() - pd.Timedelta(days=600),
                             end=price_idx_naive.max(), freq="D")
div_daily_c = divs_tk.reindex(all_cal_c, fill_value=0.0).fillna(0.0)

ttm_prev_365 = div_daily_c.rolling(365).sum().shift(365).reindex(price_idx_naive).copy()
ttm_prev_365.index = t_price.index

cal_ttm_prev = pd.Series(np.nan, index=price_idx_naive, dtype=float)
for i, dt in enumerate(price_idx_naive):
    hi = dt - pd.DateOffset(months=12)
    lo = dt - pd.DateOffset(months=24)
    cal_ttm_prev.iloc[i] = divs_tk[(divs_tk.index > lo) & (divs_tk.index <= hi)].sum()
cal_ttm_prev.index = t_price.index

growth_365 = (
    (ttm_td / ttm_prev_365.where(ttm_prev_365 > 0.001)) - 1.0
).where(is_payer).clip(GROWTH_CLIP_LO, GROWTH_CLIP_HI) * 100

cal_ttm_now = ttm_variants["12-mo calendar"]
is_pay_cal  = cal_ttm_now > 0.001
growth_cal  = (
    (cal_ttm_now / cal_ttm_prev.where(cal_ttm_prev > 0.001)) - 1.0
).where(is_pay_cal).clip(GROWTH_CLIP_LO, GROWTH_CLIP_HI) * 100

# Build smoothed variants: EWMA applied to growth_365 for several spans
growth_variants = {"365d (raw)": growth_365, "12-mo calendar (raw)": growth_cal}
for span in [10, 20, 30, 60, 90, 120]:
    growth_variants[f"365d EWMA-{span}d"] = (
        growth_365.ffill().ewm(span=span, adjust=False).mean()
        .where(is_payer)   # mask back to payer days
    )

# -- Yield rank (40% component): load pre-computed cross-sectional rank --
yield_rank_path = DATA_DIR / "yield_rank.parquet"
t_yrank = None
if yield_rank_path.exists():
    yrank_df = pd.read_parquet(yield_rank_path)
    if hasattr(yrank_df.index, "tz") and yrank_df.index.tz is not None:
        yrank_df.index = yrank_df.index.tz_localize(None)
    if TICKER in yrank_df.columns:
        t_yrank = yrank_df[TICKER].dropna().reindex(t_price.index, method="ffill").dropna()

# -- Charts --
grw_df = pd.DataFrame({"time": growth_365.index, "value": growth_365.values})

grw_chart = (
    Chart(height=210)
    .baseline(grw_df, name="YoY div growth %", color="#1976d2")
    .smoothing_control(
        variants=growth_variants,
        label="Growth method",
        color="#1976d2",
        variants_init="365d EWMA-60d",
    )
)

panes  = [grw_chart]
titles = [f"{TICKER} — Component 2: YoY Dividend Growth %  (slider: raw → EWMA smoothed)"]

if t_yrank is not None:
    yrk_df = pd.DataFrame({"time": t_yrank.index, "value": t_yrank.values})
    panes.append(Chart(height=150).area(yrk_df, name="Yield rank (0=worst, 1=best)", color="#26a69a"))
    titles.append(f"{TICKER} — Yield Rank (cross-sectional, 40% weight)")

Dashboard(panes=panes, titles=titles, theme="dark")
```

## Vol Control Regime — How It Works

The **vol-control overlay** (`strategy_vol_control.py`) adds a market-stress gate on top of the dividend signal.
When the gate is **closed** (red shading in Panel 5 below), the portfolio holds zero positions and sits in cash.

**Computation pipeline — 4 steps:**

| Step | Operation | Output shape |
|------|-----------|-------------|
| 1 | 21-day rolling realised vol per instrument, annualised: `std(ret, 21) × √252` | `date × ticker` |
| 2 | Cross-sectional **median** across ~3,997 LSE Common Stocks → single market-vol proxy | `date` (scalar) |
| 3 | 252-day rolling **z-score** of that scalar: `(mkt_vol − μ₂₅₂) / σ₂₅₂` | `date` (scalar) |
| 4 | `regime_ok = z < threshold (2.5)` → False = go flat at next monthly rebalance | boolean |

**ETFs are excluded before computing vol.**
The raw `lse_prices.parquet` contains ~7,174 LSE tickers including 3,177 ETFs (iShares, WisdomTree, GraniteShares leveraged products).
These are excluded using the `lse_equity_tickers.txt` allowlist built from `eodhd.exchange_tickers WHERE type = 'Common Stock'`.
A single ETF bad-data tick (e.g. `3SVP`: 0.10 → 2335 GBX in one day, +2,334,900%) dragged the **mean** vol from ~13% to ~8,456% and corrupted the z-score window for an entire year.

**Why cross-sectional median vol (not mean)?**
Even after ETF exclusion, median is robust to any residual data errors in the ~3,997 remaining tickers.
A single bad close price inflates that ticker’s 21-day vol — median of 3,997 tickers is completely immune.

**Why z-score and not raw vol?**
The regime of UK equity vol shifted structurally post-2020. A fixed raw-vol threshold (e.g. 20%) would fire too often in high-vol regimes and too rarely in low-vol ones. The z-score normalises against the trailing 252-day distribution, making the threshold self-adapting.

**Threshold = 2.5 (chosen by sweep):**
Sweep of 0.5–2.5 showed max DD plateaus at −14.2% for thresholds ≥1.5 (same crash exits). At 2.5, CAGR is highest (15.2%) with identical DD protection — it avoids false exits during vol spikes that do **not** lead to crashes.


```python

# ── Vol Regime: step-by-step recomputation from raw prices ───────────────────
VOL_Z_THRESHOLD   = 2.5
VOL_LOOKBACK      = 21    # days for realised vol
Z_WINDOW          = 252   # days for rolling z-score normalisation
MIN_VALID_TICKERS = 100   # suppress thin-universe early dates

# Load and pivot price data (prices_long already in scope from cell 2)
prices_wide = prices_long.pivot_table(index="date", columns="ticker", values="close")
prices_wide.index = pd.to_datetime(prices_wide.index).normalize()
prices_wide = prices_wide.sort_index()

# Restrict to Common Stock universe (eodhd.exchange_tickers type='Common Stock').
# The allowlist is generated once by sfera/_tmp_save_equity_list.py from eodhd.exchange_tickers.
# This properly excludes all ETFs (GraniteShares 3x, WisdomTree 5x, iShares, etc.)
# without incorrectly dropping European common stocks with digit-prefixed tickers
# (e.g. 0A00=Akzo Nobel, 0A1U=Uber).
_equity_list_path = SHARED_DATA / "lse_equity_tickers.txt"
with open(_equity_list_path) as _f:
    _equity_tickers = set(line.strip() for line in _f if line.strip())

equity_cols = [c for c in prices_wide.columns if c in _equity_tickers]
n_excluded = len(prices_wide.columns) - len(equity_cols)
print(f"Universe: {len(equity_cols)} equity tickers  ({n_excluded} ETFs / other types excluded)")
prices_wide = prices_wide[equity_cols]

# Step 1 — 21d realised vol per instrument (annualised)
daily_ret = prices_wide.pct_change(fill_method=None)
vol_21    = daily_ret.rolling(VOL_LOOKBACK).std() * np.sqrt(252)

# Step 2 — cross-sectional MEDIAN (robust against bad data / remaining outliers)
# Using median not mean: a single bad-data ticker (e.g. 3SVP: 0.1→2335 GBX in one day)
# drags the mean vol to 8456% and corrupts the z-score window for 252 days.
n_valid = vol_21.notna().sum(axis=1)
mkt_vol = vol_21.median(axis=1).where(n_valid >= MIN_VALID_TICKERS)

sparse_days = int((n_valid < MIN_VALID_TICKERS).sum())
if sparse_days:
    first_good = n_valid[n_valid >= MIN_VALID_TICKERS].index[0]
    print(f"\u26a0  Masked {sparse_days} thin-universe days. First usable date: {first_good.date()}")

# Step 3 — 252d rolling z-score
roll_mean = mkt_vol.rolling(Z_WINDOW, min_periods=63).mean()
roll_std  = mkt_vol.rolling(Z_WINDOW, min_periods=63).std()
mkt_vol_z = (mkt_vol - roll_mean) / roll_std

# Step 4 — regime gate (True = invested, False = cash)
regime_ok    = (mkt_vol_z < VOL_Z_THRESHOLD).astype(float)
cash_periods = (regime_ok == 0)

# ── Selected ticker's own vol (for comparison) ───────────────────────────────
t_vol = vol_21[TICKER].dropna() if TICKER in vol_21.columns else pd.Series(dtype=float, name=TICKER)

mkt_vol_pct = mkt_vol * 100
t_vol_pct   = t_vol   * 100

def _df(s):
    return pd.DataFrame({"time": s.index, "value": s.values})

# Constant threshold lines as series (Chart has no hline method)
_idx = mkt_vol_z.dropna().index
thr_hi_df = pd.DataFrame({"time": _idx, "value":  VOL_Z_THRESHOLD})
thr_lo_df = pd.DataFrame({"time": _idx, "value": -VOL_Z_THRESHOLD})

# Regime shade overlay for price chart (0=cash=shade)
regime_for_shade = regime_ok.reindex(t_price.index, method="ffill").fillna(1)
shade_df   = pd.DataFrame({"time": regime_for_shade.index, "position": regime_for_shade.astype(int)})
t_price_df = pd.DataFrame({"time": t_price.index, "value": t_price.values})

n_cash_days  = int(cash_periods.sum())
pct_cash     = n_cash_days / len(regime_ok) * 100
crisis_dates = mkt_vol_z[mkt_vol_z > VOL_Z_THRESHOLD].index
print(f"Cash days (vol filter fired): {n_cash_days} ({pct_cash:.1f}% of trading days)")
if len(crisis_dates):
    print(f"First stress date: {crisis_dates[0].date()}  |  Last: {crisis_dates[-1].date()}")

# ── Charts ────────────────────────────────────────────────────────────────────
price_chart = Chart(height=200).line(t_price_df, name=f"{TICKER} close (GBX)", color="#90caf9")
if shade_df["position"].eq(0).any():
    price_chart.shade(shade_df, position_col="position", color="#ef5350")

Dashboard(
    panes=[
        # Panel 1: cross-sectional MEDIAN realised vol (% annualised)
        Chart(height=180)
            .line(_df(mkt_vol_pct), name="Median realised vol \u2014 LSE equities (ann %)", color="#ff9800"),

        # Panel 2: 252d rolling z-score with threshold lines
        Chart(height=160)
            .baseline(_df(mkt_vol_z), base_value=0, value_col="value")
            .line(thr_hi_df, name=f"threshold +{VOL_Z_THRESHOLD}", color="#ef5350")
            .line(thr_lo_df, name=f"threshold \u2212{VOL_Z_THRESHOLD}", color="#26a69a"),

        # Panel 3: regime gate (1 = invested, 0 = cash)
        Chart(height=80)
            .area(_df(regime_ok), name="Regime OK (1=invested 0=cash)", color="#26a69a"),

        # Panel 4: ticker's own vol vs universe median
        Chart(height=160)
            .line(_df(t_vol_pct),   name=f"{TICKER} realised vol (ann %)", color="#7986cb")
            .line(_df(mkt_vol_pct), name="Universe median vol",            color="#ff9800"),

        # Panel 5: price with red shading when vol gate is active
        price_chart,
    ],
    titles=[
        f"Market Vol Proxy  \u00b7  MEDIAN 21d realised vol  \u00b7  {len(equity_cols)} equities (ETFs excluded)  \u00b7  median {mkt_vol_pct.median():.1f}%",
        f"Vol Z-Score (252d)  \u00b7  threshold = {VOL_Z_THRESHOLD}  \u00b7  cash {pct_cash:.1f}% of days",
        f"Regime Gate  \u00b7  1 = long  |  0 = flat (cash)",
        f"{TICKER} own vol vs universe median",
        f"{TICKER} price (GBX)  \u00b7  red = cash (vol gate fired)",
    ],
    theme="dark",
)
```
