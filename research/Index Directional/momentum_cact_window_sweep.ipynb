# CACT Momentum Timing Threshold Research

Research notebook for the **CAC 40 TR time-series momentum** signal that runs in production via
`sfera/signals/cact_momentum.py`.

## Production signal (locked parameters)

```
signal_T = 1  if  log(close_T / close_{T-139}) >= 0.005  else  0
position_{T+1} = signal_T          # MOC execution next bar (signal_delay = 1)
```

| Parameter | Value | Meaning |
|-----------|-------|---------|
| `WINDOW`  | **139** trading days |  7-month lookback |
| `THETA`   | **0.005** log-return | +0.50% threshold |
| Data      | `bbgidx.index_total_return` (`CACT`) | CAC 40 Gross Return Index |
| Execution | MOC (market-on-close), next day | Signal computed from close at 06:00 CET |
| Output    | `sfera/data/signals/cact_momentum.parquet` | columns: `ticker`, `date`, `signal`, `momentum` |

````python
import os, sys, pathlib, warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import sfera_db
from signum import Chart, Dashboard


# Paths notebook lives at btest/research/Index Directional/
_HERE       = pathlib.Path(".").resolve()          # btest/research/Index Directional/
_BTEST_ROOT = _HERE.parents[1]                     # btest/
_WS_ROOT    = _BTEST_ROOT.parent                   # workspace root

for _p in [
    str(_WS_ROOT),
    str(_WS_ROOT / "sfera-db"),
    str(_WS_ROOT / "signum"),
    str(_BTEST_ROOT / "src"),
]:
    if _p not in sys.path:
        sys.path.insert(0, _p)


def _q(sql):
    return (sfera_db.query(sql)
            .assign(date=lambda d: pd.to_datetime(d["date"]))
            .set_index("date"))

cactr = _q("SELECT trade_date AS date, open_price AS open, close_price AS close "
           "FROM bbgidx.index_total_return WHERE ticker = 'CACT' ORDER BY trade_date")

# Keep full history where index data exists
df = cactr[["open", "close"]].copy()
df["log_ret"] = np.log(df["close"] / df["close"].shift(1))
df = df.dropna().copy()  

print(f"Data: {df.index[0].date()} -> {df.index[-1].date()} ({len(df)} days)")
````

````python
## 1) Momentum window explorer

def _to_lc(series, col_name="value"):
    return series.rename(col_name).reset_index().rename(columns={"date": "time"})

def _momentum(s, w):
    # Log momentum over lookback window
    return np.log(s / s.shift(w))

WINDOWS = sorted(set(list(range(20, 253, 7)) + [126]))
mom_by_win = {w: _momentum(df["close"], w).dropna() for w in WINDOWS}

_first = min(s.index[0] for s in mom_by_win.values())
df_v = df.loc[_first:].copy()
for w in WINDOWS:
    mom_by_win[w] = mom_by_win[w].loc[_first:]

p0 = Chart(theme="dark", height=200).line(_to_lc(df_v["close"]), name="CAC TR", color="#9aa6ff")

p1 = (Chart(theme="dark", height=220)
    .line(_to_lc(mom_by_win[126]), name="momentum", color="#2a9d8f")
    .smoothing_control(
        variants={w: mom_by_win[w] for w in WINDOWS},
        variants_init=126,
        label="mom-window",
        color="#2a9d8f",
    )
)

Dashboard(
    panes=[p0, p1],
    titles=["CAC Total Return", "Log Momentum (window sweep)"],
    theme="dark",
).show()
````

````python
## 4) Momentum gate prep (non-Signum) -- full history from 2001

MOM_WINDOW = 139
THRESHOLD  = 0.005
THETA_STEP = 0.0025
START_DATE = "2015-01-01"

mom = _momentum(df["close"], MOM_WINDOW)
sig = mom.dropna()
sig = sig.loc[sig.index >= START_DATE]

df_c      = df.loc[sig.index[0]:].copy()
sig       = sig.loc[df_c.index]
daily_ret = np.exp(df_c["log_ret"]) - 1

theta_min = float(np.floor((sig.min() - THETA_STEP) / THETA_STEP) * THETA_STEP)
theta_max = float(np.ceil((sig.max()  + THETA_STEP) / THETA_STEP) * THETA_STEP)

gate0      = (sig >= THRESHOLD).astype(float)
strat_ret0 = daily_ret * gate0.shift(1).fillna(0)
eq_strat0  = (1 + strat_ret0).cumprod()
eq_bnh     = (1 + daily_ret).cumprod()

# Exact geometric stats
_n_years = len(eq_strat0) / 252
_cagr    = eq_strat0.iloc[-1] ** (1 / _n_years) - 1
_bh_cagr = eq_bnh.iloc[-1]   ** (1 / _n_years) - 1
_vol     = strat_ret0.std() * np.sqrt(252)
_sr      = _cagr / _vol if _vol > 0 else 0
_mdd     = (eq_strat0 / eq_strat0.cummax() - 1).min()
print(f"Sample : {df_c.index[0].date()} -> {df_c.index[-1].date()}  ({len(df_c)} days)")
print(f"B&H    : {eq_bnh.iloc[-1] - 1:.1%}  CAGR={_bh_cagr*100:+.1f}%")
print(f"theta={THRESHOLD}  CAGR={_cagr*100:+.1f}%  Vol={_vol*100:.1f}%  SR={_sr:.2f}  MaxDD={_mdd*100:.1f}%  Return={eq_strat0.iloc[-1]-1:.1%}")
````

````python
## 5) Signum view (concise)

p_eq = (Chart(theme="dark", height=230)
    .line(_to_lc(eq_strat0), name=f"Momentum w={MOM_WINDOW} theta={THRESHOLD}", color="#2a9d8f")
    .line(_to_lc(eq_bnh),    name="B&H CACT",                                  color="#666666")
)

p_sig = (Chart(theme="dark", height=170)
    .baseline(_to_lc(sig), base_value=THRESHOLD,
              name=f"Momentum({MOM_WINDOW}) raw",
              topLineColor="#2a9d8f",
              topFillColor1="rgba(42,157,143,0.10)", topFillColor2="transparent",
              bottomLineColor="rgba(140,140,140,0.7)",
              bottomFillColor1="rgba(140,140,140,0.05)", bottomFillColor2="transparent")
)

(Dashboard(
    panes=[p_eq, p_sig],
    titles=[
        f"Equity | momentum({MOM_WINDOW}) >= theta  [{START_DATE}]",
        "Signal | baseline = theta",
    ],
    theme="dark",
    execution=1,
).threshold_control(
    df=sig.to_frame("value"),
    signal_mode=">=",
    threshold=THRESHOLD,
    min_val=theta_min,
    max_val=theta_max,
    step=THETA_STEP,
    price_pane=0,
    prices=df_c[["open", "close"]],
    equity_pane=0,
).show())
````

````python
## Trade-level PnL audit -- $100 start, LAG-1 MOC
# Execution: signal[T] computed from close[T] -> enter at MOC = close[T] -> hold on T+1
# Therefore: pos[T] = signal[T-1]   (position today was set by YESTERDAY's signal)
# avg_cost = close[T-1] = the MOC fill price, carried while pos==1

INITIAL_NAV = 100.0

_mom_raw = np.log(df["close"] / df["close"].shift(MOM_WINDOW))
_sig_raw  = (_mom_raw >= THRESHOLD).astype(int)

audit = pd.DataFrame({
    "open"     : df["open"],
    "close"    : df["close"],
    "mom"      : _mom_raw,
    "signal"   : _sig_raw,                               # 0/1: today's signal (from today's close)
    "pos"      : _sig_raw.shift(1).fillna(0).astype(int), # 0/1: position held TODAY = yesterday's signal
    "ret_cc"   : np.exp(df["log_ret"]) - 1,
}).dropna()
audit = audit.loc[START_DATE:].copy()

# Entry / exit flags
_prev_pos = audit["pos"].shift(1).fillna(0).astype(int)
_is_entry = (audit["pos"] == 1) & (_prev_pos == 0)
_is_exit  = (audit["pos"] == 0) & (_prev_pos == 1)

# avg_cost = close of the signal day (i.e. close[T-1] from today's view) -- the MOC fill price
audit["avg_cost"]  = (audit["close"].shift(1)
                      .where(_is_entry)
                      .ffill()
                      .where(audit["pos"] == 1))

# Dollar P&L
audit["strat_ret"] = audit["ret_cc"] * audit["pos"]
audit["nav"]       = INITIAL_NAV * (1 + audit["strat_ret"]).cumprod()
audit["bh_nav"]    = INITIAL_NAV * (1 + audit["ret_cc"]).cumprod()
audit["daily_pnl"] = audit["nav"].diff().fillna(0)

# Show rows around last 10 transitions + first/last 5
_changes  = audit.index[_is_entry | _is_exit]
_show_idx = sorted(set(
    list(audit.index[:5]) +
    [idx2 for c in _changes[-10:]
     for idx2 in audit.index[max(0, audit.index.get_loc(c)-1):
                              audit.index.get_loc(c)+3].tolist()] +
    list(audit.index[-5:])
))
_cols = ["open", "close", "mom", "signal", "pos", "avg_cost", "ret_cc", "strat_ret", "daily_pnl", "nav", "bh_nav"]
_show = audit.loc[_show_idx, _cols]

pd.set_option("display.float_format", lambda x: f"{x:+.4f}" if abs(x) < 100 else f"{x:.2f}")
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 20)

print("KEY: pos = signal from PREVIOUS day (LAG-1). So pos=1 when signal was 1 yesterday.")
print("     avg_cost = MOC fill price (close of entry day), carried for duration of trade.")
print()
print(_show.to_string())

_n = len(audit) / 252
print(f"\n{'â”€'*90}")
print(f"Strategy : NAV={audit['nav'].iloc[-1]:.2f}  total={audit['nav'].iloc[-1]/INITIAL_NAV-1:.1%}  CAGR={(audit['nav'].iloc[-1]/INITIAL_NAV)**(1/_n)-1:+.2%}")
print(f"B&H      : NAV={audit['bh_nav'].iloc[-1]:.2f}  total={audit['bh_nav'].iloc[-1]/INITIAL_NAV-1:.1%}  CAGR={(audit['bh_nav'].iloc[-1]/INITIAL_NAV)**(1/_n)-1:+.2%}")
print(f"Days long: {int(audit['pos'].sum())}/{len(audit)} ({audit['pos'].mean():.1%})")
````

````python
## 7) Execution-policy cross-validation -- Python apply_execution() vs manual DataFrame
# Proves that Signum execution=0 and 'next_close' produce IDENTICAL
# daily returns to the manually-computed diagnostic DataFrame at each lag.
# Also verifies the corrected Signum Sharpe (CAGR / sample_vol) matches Python.

from signum.engine.chart import Chart as _Chart

TEST_START = "2022-01-01"
TEST_END   = "2022-12-31"

# Full-series computations: apply shifts to whole history, then slice -- no boundary NaN
_mom_f = _momentum(df["close"], MOM_WINDOW)
_ret_f = np.exp(df["log_ret"]) - 1

_gate0_full = (_mom_f.shift(1) >= THRESHOLD).astype(float)   # LAG-0  pos[T]=mom[T-1]>=theta
_gate1_full = (_mom_f.shift(2) >= THRESHOLD).astype(float)   # LAG-1  pos[T]=mom[T-2]>=theta

_gate0 = _gate0_full.loc[TEST_START:TEST_END]
_gate1 = _gate1_full.loc[TEST_START:TEST_END]
_ret   = _ret_f.loc[TEST_START:TEST_END]

# --- Manual returns (reference) ---
r0_manual = _gate0 * _ret                           # LAG-0  (same_bar)
r1_manual = _gate1 * _ret                           # LAG-1  (next_close / MOC)

# --- Signum apply_execution() ---
# same_bar   => signal.fillna(0) * ret              (lag=0)
# next_close => signal.shift(1).fillna(0) * ret     (lag=1)
# For LAG-1: pass gate0 starting at TEST_START so shift(1) captures the boundary bar
# (gate0 is already full-history sliced so no info is lost at the window edge)
r0_signum = _Chart.apply_execution(_gate0, _ret, execution=0)
_gate0_ext = _gate0_full.loc[TEST_START:]
r1_signum  = (
    _Chart.apply_execution(_gate0_ext, _ret_f.loc[TEST_START:], execution=1)
    .loc[TEST_START:TEST_END]
)

diff0 = (r0_manual - r0_signum).abs()
diff1 = (r1_manual - r1_signum).abs()

_W = 72
print("=" * _W)
print(f"  Execution cross-validation  |  {TEST_START} to {TEST_END}  |  W={MOM_WINDOW}  theta={THRESHOLD}")
print("=" * _W)
_hdr = f"\n{'Policy':<29} {'Manual total':>13} {'Signum total':>13} {'Max|diff|':>12}  Match"
print(_hdr)
print("-" * _W)

for _label, _r_py, _r_sg, _diff in [
    ("same_bar   (LAG-0, lookahead)", r0_manual, r0_signum, diff0),
    ("next_close (LAG-1, MOC true)", r1_manual, r1_signum, diff1),
]:
    _py_tot = (1 + _r_py).prod() - 1
    _sg_tot = (1 + _r_sg).prod() - 1
    _ok     = "EXACT" if _diff.max() < 1e-12 else f"MISMATCH  max={_diff.max():.3e}"
    print(f"{_label:<29} {_py_tot:>+12.4%} {_sg_tot:>+12.4%} {_diff.max():>12.3e}   {_ok}")

# --- First 12 rows side-by-side ---
print("\n-- First 12 rows: LAG-0 (same_bar) --------------------------------------")
_cmp0 = pd.DataFrame({
    "mom[T-1]": _mom_f.shift(1).loc[TEST_START:TEST_END],
    "gate_py":  _gate0.astype(int),
    "gate_sg":  _gate0.astype(int),
    "ret":      _ret,
    "r_manual": r0_manual,
    "r_signum": r0_signum,
    "diff":     diff0,
}).head(12)
print(_cmp0.to_string())

print("\n-- First 12 rows: LAG-1 MOC (next_close) --------------------------------")
_cmp1 = pd.DataFrame({
    "mom[T-2]": _mom_f.shift(2).loc[TEST_START:TEST_END],
    "gate_py":  _gate1.astype(int),
    "gate_sg":  _gate0_full.shift(1).loc[TEST_START:TEST_END].fillna(0).astype(int),
    "ret":      _ret,
    "r_manual": r1_manual,
    "r_signum": r1_signum,
    "diff":     diff1,
}).head(12)
print(_cmp1.to_string())

# --- Sharpe formula verification: corrected JS == Python cell 5 ---
print("\n-- Sharpe formula check (full history from 2001) -------------------------")
print("   Python = CAGR / (strat_ret.std(ddof=1) * sqrt(252))")
print("   Signum = CAGR / sqrt((sumR2 - sumR^2/N) / (N-1) * 252)  [now fixed]")
print()
for _lbl, _gate in [("same_bar   (LAG-0)", _gate0_full), ("next_close (LAG-1)", _gate1_full)]:
    _r      = (_ret_f * _gate).loc["2001-01-01":]
    _nav    = (1 + _r).prod()
    _n_y    = len(_r) / 252
    _cagr   = _nav ** (1 / _n_y) - 1
    _vol_py = _r.std(ddof=1) * np.sqrt(252)
    _sr_py  = _cagr / _vol_py if _vol_py > 0 else 0
    _N, _sR, _sR2 = len(_r), _r.sum(), (_r**2).sum()
    _vol_js = np.sqrt((_sR2 - _sR**2 / _N) / (_N - 1) * 252) if _N > 1 else 1.0
    _sr_js  = _cagr / _vol_js if _vol_js > 0 else 0
    _match  = "MATCH" if abs(_sr_py - _sr_js) < 1e-10 else "MISMATCH"


````
