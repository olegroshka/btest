Separate notebook to test medium-term momentum with interactive window and threshold controls.

```python
import os, sys, pathlib, warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd

# Paths
_HERE       = pathlib.Path(".").resolve()
_IDX_DIR    = _HERE.parent
_BTEST_ROOT = _HERE.parents[2]
_WS_ROOT    = _BTEST_ROOT.parent

for _p in [
    str(_HERE),
    str(_WS_ROOT),
    str(_WS_ROOT / "sfera-db"),
    str(_WS_ROOT / "signum"),
    str(_BTEST_ROOT / "src"),
]:
    if _p not in sys.path:
        sys.path.insert(0, _p)

import sfera_db
from signum import Chart, Dashboard

print("Imports OK")
```

```python
def _q(sql):
    return (sfera_db.query(sql)
            .assign(date=lambda d: pd.to_datetime(d["date"]))
            .set_index("date"))

cactr = _q("SELECT trade_date AS date, close_price AS close "
           "FROM bbgidx.index_total_return WHERE ticker = 'CACT' ORDER BY trade_date")

# Keep full history where index data exists
df = cactr[["close"]].copy()
df["log_ret"] = np.log(df["close"] / df["close"].shift(1))
df = df.loc[df.index >= "2000-01-01"].dropna().copy()

print(f"Data: {df.index[0].date()} -> {df.index[-1].date()} ({len(df)} days)")
```

```python
## 1) Momentum window explorer

def _to_lc(series, col_name="value"):
    return series.rename(col_name).reset_index().rename(columns={"date": "time"})

def _momentum(s, w):
    # Log momentum over lookback window
    return np.log(s / s.shift(w))

WINDOWS = sorted(set(list(range(20, 253, 7)) + [126]))
mom_by_win = {w: _momentum(df["close"], w).dropna() for w in WINDOWS}

_first = min(s.index[0] for s in mom_by_win.values())
df_v = df.loc[_first:].copy()
for w in WINDOWS:
    mom_by_win[w] = mom_by_win[w].loc[_first:]

p0 = Chart(theme="dark", height=200).line(_to_lc(df_v["close"]), name="CAC TR", color="#9aa6ff")

p1 = (Chart(theme="dark", height=220)
    .line(_to_lc(mom_by_win[126]), name="momentum", color="#2a9d8f")
    .smoothing_control(
        variants={w: mom_by_win[w] for w in WINDOWS},
        variants_init=126,
        label="mom-window",
        color="#2a9d8f",
    )
)

Dashboard(
    panes=[p0, p1],
    titles=["CAC Total Return", "Log Momentum (window sweep)"],
    theme="dark",
).show()
```

```python
## 2) 126d momentum gate with threshold slider

MOM_WINDOW = 140
THRESHOLD = 0.00
THETA_STEP = 0.0025

mom = _momentum(df_v["close"], MOM_WINDOW)
sig = mom.shift(1).dropna()  # lagged signal for no lookahead

valid_start = sig.index[0]
df_c = df_v.loc[valid_start:].copy()
sig = sig.loc[valid_start:]

# threshold_control compounds as (1 + r), so feed it SIMPLE returns, not log returns
daily_ret = np.exp(df_c["log_ret"]) - 1

# Dynamic slider bounds so top end can always replicate full-invested if desired
theta_min = float(np.floor((sig.min() - THETA_STEP) / THETA_STEP) * THETA_STEP)
theta_max = float(np.ceil((sig.max() + THETA_STEP) / THETA_STEP) * THETA_STEP)

gate0 = (sig >= THRESHOLD).astype(float)
eq_strat0 = (1 + daily_ret * gate0).cumprod()
eq_bnh = (1 + daily_ret).cumprod()

p_eq = (Chart(theme="dark", height=230)
    .line(_to_lc(eq_strat0), name="Strategy", color="#2a9d8f")
    .line(_to_lc(eq_bnh), name="B&H", color="#666666")
)

p_sig = (Chart(theme="dark", height=170)
    .baseline(_to_lc(sig), base_value=THRESHOLD,
              name=f"Momentum({MOM_WINDOW}) lag1",
              topLineColor="#2a9d8f",
              topFillColor1="rgba(42,157,143,0.10)", topFillColor2="transparent",
              bottomLineColor="rgba(140,140,140,0.7)",
              bottomFillColor1="rgba(140,140,140,0.05)", bottomFillColor2="transparent")
)

(Dashboard(
    panes=[p_eq, p_sig],
    titles=[
        f"Equity | gate: momentum({MOM_WINDOW}, lag1) >= theta",
        f"Signal | baseline = theta",
    ],
    theme="dark",
).threshold_control(
    df=sig.to_frame("value"),
    signal_mode=">=",
    threshold=THRESHOLD,
    min_val=theta_min,
    max_val=theta_max,
    step=THETA_STEP,
    price_pane=0,
    daily_returns=daily_ret,
    bh_returns=daily_ret,
    equity_pane=0,
).show())

print(f"theta range: [{theta_min:.4f}, {theta_max:.4f}]")
print(f"Static B&H total return on this chart sample: {eq_bnh.iloc[-1] - 1:.1%}")
```

```python
## 3) Sweep momentum window Ã— threshold

# Two live sliders would not be correct with current signum wiring, because the threshold
# control does not know which momentum-window variant is currently selected.
# So do the proper thing: evaluate the full 2D grid in Python.

SWEEP_WINDOWS = sorted(set(list(range(20, 253, 7)) + [63, 84, 126, 168, 189, 252]))
THETA_GRID = np.round(np.arange(-0.20, 0.205, 0.005), 4)

# Fair comparison: use a common sample start across all windows after lagging the signal.
sig_by_win = {w: _momentum(df["close"], w).shift(1) for w in SWEEP_WINDOWS}
common_start = max(sig.dropna().index[0] for sig in sig_by_win.values())
df_sweep = df.loc[common_start:].copy()
ret_sweep = df_sweep["log_ret"].dropna()

rows = []
for w in SWEEP_WINDOWS:
    sig_w = sig_by_win[w].loc[ret_sweep.index].dropna()
    aligned_idx = ret_sweep.index.intersection(sig_w.index)
    sig_w = sig_w.loc[aligned_idx]
    ret_w = ret_sweep.loc[aligned_idx]

    for theta in THETA_GRID:
        gate = (sig_w >= theta).astype(float)
        strat = ret_w * gate

        ann = strat.mean() * 252
        vol = strat.std() * np.sqrt(252)
        sharpe = ann / vol if vol > 0 else np.nan
        total = np.exp(strat.sum()) - 1
        bnh_total = np.exp(ret_w.sum()) - 1
        cagr = np.exp(strat.mean() * 252) - 1
        max_dd = (np.exp(strat.cumsum()) / np.exp(strat.cumsum()).cummax() - 1).min()

        rows.append({
            "window": w,
            "theta": float(theta),
            "total_return": total,
            "cagr": cagr,
            "vol": vol,
            "sharpe": sharpe,
            "max_dd": max_dd,
            "in_mkt": gate.mean(),
            "days": len(strat),
            "bnh_total": bnh_total,
        })

sweep = pd.DataFrame(rows).sort_values(["sharpe", "total_return"], ascending=[False, False]).reset_index(drop=True)

print(f"Common sweep sample: {ret_sweep.index[0].date()} -> {ret_sweep.index[-1].date()} ({len(ret_sweep)} days)")
print()
print("Top 15 by Sharpe")
display(
    sweep[["window", "theta", "sharpe", "cagr", "vol", "max_dd", "in_mkt", "total_return"]]
    .head(15)
    .style.format({
        "theta": "{:.3f}",
        "sharpe": "{:.2f}",
        "cagr": "{:.2%}",
        "vol": "{:.2%}",
        "max_dd": "{:.2%}",
        "in_mkt": "{:.1%}",
        "total_return": "{:.1%}",
    })
)

print("Top 15 by Total Return")
display(
    sweep.sort_values(["total_return", "sharpe"], ascending=[False, False])
    [["window", "theta", "total_return", "cagr", "sharpe", "vol", "max_dd", "in_mkt"]]
    .head(15)
    .style.format({
        "theta": "{:.3f}",
        "total_return": "{:.1%}",
        "cagr": "{:.2%}",
        "sharpe": "{:.2f}",
        "vol": "{:.2%}",
        "max_dd": "{:.2%}",
        "in_mkt": "{:.1%}",
    })
)

sharpe_pivot = sweep.pivot_table(index="window", columns="theta", values="sharpe")
ret_pivot = sweep.pivot_table(index="window", columns="theta", values="total_return")

best_sharpe = sweep.iloc[0]
best_return = sweep.sort_values(["total_return", "sharpe"], ascending=[False, False]).iloc[0]
print()
print(f"Best Sharpe: window={int(best_sharpe['window'])}, theta={best_sharpe['theta']:.3f}, Sharpe={best_sharpe['sharpe']:.2f}, CAGR={best_sharpe['cagr']:.2%}, InMkt={best_sharpe['in_mkt']:.1%}")
print(f"Best Return: window={int(best_return['window'])}, theta={best_return['theta']:.3f}, TotalReturn={best_return['total_return']:.1%}, Sharpe={best_return['sharpe']:.2f}, InMkt={best_return['in_mkt']:.1%}")
```
